```python
import pymorphy3
import re
```

```python
morph = pymorphy3.MorphAnalyzer()

print("Введите текст:")
text = input()

sentences = re.split(r'[.!?]+', text) #считаем предложения
sentences = [s.strip() for s in sentences if s.strip()]

print(f"Всего предложений: {len(sentences)}")
print()

#показываем все предложения
for i, sentence in enumerate(sentences, 1):
    print(f"Предложение {i}: {sentence}")

#чистим от пунктуации
def clean_word(word):
    clean = re.sub(r'^[^\wА-Яа-яЁё]+|[^\wА-Яа-яЁё]+$', '', word)
    clean = clean.lower()
    return clean

#делим на слова
words = text.split()
clean_words = []

#чистим каждое слово
for word in words:
    clean = clean_word(word)
    if clean:
        clean_words.append(clean)

print(f"Очищенные слова: {clean_words}")
```

```python
#анализируем слово
for word in clean_words:
    parsed = morph.parse(word)[0]

    tag_str = str(parsed.tag)

    if parsed.tag.POS:
        pos = str(parsed.tag.POS)
    elif ',' in tag_str:
        pos = tag_str.split(',')[0]


    print(f"Слово: '{word}'")
    print(f"  Лемма: {parsed.normal_form}")
    print(f"  Часть речи: {pos}")

    #число, если есть
    if 'sing' in tag_str:
        print(f"  Число: единственное")
    elif 'plur' in tag_str:
        print(f"  Число: множественное")

    print()
```

```python
#статистика по числам
singular_count = 0
plural_count = 0
no_number_count = 0

for word in clean_words:
    parsed = morph.parse(word)[0]
    tag_str = str(parsed.tag)

    #проверяем
    if 'sing' in tag_str:
        singular_count += 1
    elif 'plur' in tag_str:
        plural_count += 1
    else:
        no_number_count += 1

print(f"Единственное число: {singular_count}")
print(f"Множественное число: {plural_count}")
print(f"Без категории числа: {no_number_count}")

#проценты
total_words = len(clean_words)
print(f"\nПроценты от всех слов ({total_words}):")
if total_words > 0:
    print(f"В единственном числе: {(singular_count/total_words)*100:.1f}%")
    print(f"Во множественном числе: {(plural_count/total_words)*100:.1f}%")
    print(f"Без числа: {(no_number_count/total_words)*100:.1f}%")
```
