```python
import requests
from bs4 import BeautifulSoup
import time
import sqlite3
import pandas as pd
import re
```

```python
#качаем информацию
url = "https://www.olympedia.org/countries"
response = requests.get(url)
time.sleep(2)

#парсим HTML
soup = BeautifulSoup(response.text, 'html.parser')

#находим таблицу на странице
table = soup.find('table')

#парсим её
if table:
    countries = []

    #проходим по всем строкам таблицы
    rows = table.find_all('tr')

    for row in rows[1:]:  #начинаем со второй строки (заголовок не нужен)
        #первая ссылка в строке
        link = row.find('a')
        if link and link.get('href'):
            href = link.get('href')

            #получаем данные
            if '/countries/' in href:
                country_code = href.split('/')[-1]
                country_name = link.text.strip()

                #ищем код нок (3 заглавные буквы)
                noc_code = ""
                cells = row.find_all(['td', 'th'])
                for cell in cells:
                    text = cell.text.strip()
                    if len(text) == 3 and text.isalpha() and text.isupper():
                        noc_code = text
                        break

                #добавляем страну
                countries.append({
                    'country_code': country_code,
                    'country_name': country_name,
                    'noc_code': noc_code if noc_code else "N/A",
                    'url': f"https://www.olympedia.org{href}"
                })

  #проверка
    if countries:
        for i, country in enumerate(countries[:10]):
            print(f"{i+1:2}. {country['country_name']:25}")
```

```python
#сбор детальной информации по странам

#участники (строка Olympic Games)
def get_participants(webpage):
    for table in webpage.find_all('table'):
        for row in table.find_all('tr'):
            if 'Olympic Games' in row.text:
                words = row.text.split()
                return words[-1]
    return "0"

#флагоносцы
def get_flagbearers(webpage):
    flagbearers_th = webpage.find('th', string='Flagbearers')
    if flagbearers_th:
        flagbearers_td = flagbearers_th.find_next('td')
        flagbearers = flagbearers_td.find('a').get_text(strip=True)
        return flagbearers
    return "0"

#описание
def get_description(webpage):
    first_p = webpage.find('p')
    if first_p:
        return first_p.get_text(strip=True)[:500]
    return "0"

#обрабатываем страны
for i in range(len(countries)):

    response = requests.get(countries[i]['url'])
    time.sleep(2)
    soup = BeautifulSoup(response.text, 'html.parser')

    countries[i]['participants'] = get_participants(soup)
    countries[i]['flagbearers'] = get_flagbearers(soup)
    countries[i]['description'] = get_description(soup)


#проверка
for i in range(5):
    country = countries[i]
    print(f"\n{i+1}. {country['country_name']}:")
    print(f"   Участников: {country['participants']}")
    print(f"   Флагоносцев: {country['flagbearers']}")
    print(f"   Описание: {country['description'][:100]}...")
```

```python
#сбор через расширенный поиск

base_url = "https://www.olympedia.org/athletes/search_athlete?affiliations=&bio=&birthcountry=&birthplace=&competition_type_id=&continent=&country=&date%5Bbirthday%5D=&date%5Bbirthmonth%5D=&date%5Bbirthyear%5D=&date%5Bdeathday%5D=&date%5Bdeathmonth%5D=&date%5Bdeathyear%5D=&deathcountry=&deathplace=&event_name=&gender=male&given_name=&id=&image=&medal=&role%5B%5D=non-olympic&season=summer&second_country=&sport=&sport_group=&surname=&year2=&year="

athletes = []

for page in range(1, 23):

    response = requests.get(f"{base_url}&page={page}")
    time.sleep(2)

    soup = BeautifulSoup(response.text, 'html.parser')
    table = soup.find('table', {'class': 'table-striped'})

    if not table:
        break

    rows = table.find_all('tr')[1:]

    for row in rows:
        cells = row.find_all('td')
        if len(cells) >= 6:
            #данные из ячеек таблицы
            athlete_id = cells[0].text.strip()
            name = cells[1].text.strip()
            countries = [c.strip() for c in cells[2].text.strip().split() if c.strip()]

            #годы
            era_cell = cells[5]
            era_text = era_cell.text.strip()

            years = []
            #разбиваем по возможным разделителям (длинное тире, дефис, короткое тире)
            parts = re.split(r'[—\-–]', era_text)

            for part in parts:
                part = part.strip()
                if part.isdigit() and len(part) == 4:
                    years.append(int(part))

            if years:
                start = min(years)
                end = max(years)
                career = end - start if end > start else 1
            else:
                #пробуем найти просто 4-значные числа
                year_matches = re.findall(r'\d{4}', era_text)
                if year_matches:
                    years = [int(y) for y in year_matches]
                    start = min(years)
                    end = max(years)
                    career = end - start if end > start else 1
                else:
                    start = end = career = 0

            athletes.append({
                'id': athlete_id,
                'name': name,
                'countries': countries,
                'start': start,
                'end': end,
                'career': career
            })

#проверка
if athletes:
    for i in range(min(5, len(athletes))):
        a = athletes[i]
        print(f"\n{i+1}. {a['name']} (ID: {a['id']})")
        print(f"   Страны: {a['countries']}")
        print(f"   Годы: {a['start']}-{a['end']} ({a['career']} лет)")
```

```python
#создание базы данных

conn = sqlite3.connect('olympics.db')
cursor = conn.cursor()

#таблица 1: страны
cursor.execute('''
CREATE TABLE IF NOT EXISTS countries (
    country_code TEXT PRIMARY KEY,
    country_name TEXT NOT NULL,
    noc_code TEXT,
    participants INTEGER DEFAULT 0,
    flagbearers INTEGER DEFAULT 0,
    description TEXT
)
''')

#таблица 2: спортсмены
cursor.execute('''
CREATE TABLE IF NOT EXISTS athletes (
    athlete_id TEXT PRIMARY KEY,
    name TEXT NOT NULL,
    start_year INTEGER,
    end_year INTEGER,
    career_years INTEGER DEFAULT 0
)
''')

#таблица 3: связь спортсменов и стран
cursor.execute('''
CREATE TABLE IF NOT EXISTS athlete_countries (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    athlete_id TEXT,
    country_code TEXT,
    FOREIGN KEY (athlete_id) REFERENCES athletes (athlete_id),
    FOREIGN KEY (country_code) REFERENCES countries (country_code)
)
''')
```

```python
print("=== Проверка переменной countries ===")
print(f"Тип переменной: {type(countries)}")
print(f"Длина: {len(countries) if hasattr(countries, '__len__') else 'нет длины'}")

# Показываем первые 100 символов если это строка
if isinstance(countries, str):
    print("\nПервые 100 символов:")
    print(countries[:100])
else:
    print("\nПервые 3 элемента:")
    for i in range(min(3, len(countries))):
        print(f"{i}: {countries[i]}")
```

```python
#заполнение базы данных

#заполняем таблицу стран
for country in countries:
    cursor.execute('''
    INSERT OR REPLACE INTO countries
    (country_code, country_name, noc_code, participants, flagbearers, description)
    VALUES (?, ?, ?, ?, ?, ?)
    ''', (
        country['country_code'],
        country['country_name'],
        country['noc_code'],
        int(country.get('participants', 0)),
        int(country.get('flagbearers', 0)),
        country.get('description', '')
    ))

#заполняем таблицу спортсменов
for athlete in athletes:
    cursor.execute('''
    INSERT OR REPLACE INTO athletes
    (athlete_id, name, start_year, end_year, career_years)
    VALUES (?, ?, ?, ?, ?)
    ''', (
        athlete['id'],
        athlete['name'],
        athlete['start'],
        athlete['end'],
        athlete['career']
    ))

    # 3. Заполняем связи спортсменов и стран
    for country_code in athlete['countries']:
        cursor.execute('''
        INSERT INTO athlete_countries (athlete_id, country_code)
        VALUES (?, ?)
        ''', (athlete['id'], country_code))

#сохраняем изменения
conn.commit()

print(f"✓ Добавлено: {len(countries)} стран, {len(athletes)} спортсменов")
```
